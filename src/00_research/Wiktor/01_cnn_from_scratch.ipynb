```python
import numpy as np
```

```python
from math import floor

l0_input_channels = 3

l1_params = np.random.randn(6, 5, 5, l0_input_channels)
l1_stride = 1


def conv(x: np.ndarray, params: np.ndarray, stride: int):
    # x shape: (h, w, number_of_input_channels)
    # params shape: (number_of_filters, ker_h, ker_w, number_of_input_channels)
    #
    # return shape: (new_h, new_w, number_of_filters)
    number_of_filters = params.shape[0]

    # kernels must be squares
    assert params.shape[1] == params.shape[2]
    ker_h = params.shape[1]

    assert x.shape[0] == x.shape[1]
    h = x.shape[0]

    # [(W−K)/S]+1
    new_h = floor(((h - ker_h) / stride) + 1)

    output_shape = (new_h, new_h, number_of_filters)
    output = np.zeros(output_shape)

    for filter_index, par in enumerate(params):
        for i in range(new_h):
            for j in range(new_h):
                start_position = (i * stride, j * stride)
                end_position = (start_position[0] + ker_h, start_position[1] + ker_h)
                region = x[start_position[0] : end_position[0], start_position[1] : end_position[1]]

                region_result = np.multiply(par, region)
                pixel_result = np.sum(region_result)
                output[i][j][filter_index] = pixel_result

    return output


def forward(x: np.ndarray) -> np.ndarray:
    # the image has to be symmetrical
    assert x.shape[0] == x.shape[1]

    x = conv(x, l1_params, l1_stride)
    print(x.shape)


forward(np.random.randn(28, 28, 1))
```
