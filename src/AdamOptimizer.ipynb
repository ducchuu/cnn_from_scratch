
# Adam Optimizer Class Documentation

This class implements the Adam (Adaptive Moment Estimation) optimization algorithm, which computes adaptive learning rates for each parameter by storing both the first and second moments of the gradients. It is widely used in deep learning for training models.

## Class Definition

```python
class Adam():
    def __init__(self, eta=0.01, beta1=0.9, beta2=0.999, epsilon=1e-8):
```

### Parameters
- `eta` (float, default=0.01): The learning rate.
- `beta1` (float, default=0.9): The exponential decay rate for the first moment estimate.
- `beta2` (float, default=0.999): The exponential decay rate for the second moment estimate.
- `epsilon` (float, default=1e-8): A small value to prevent division by zero in the update step.

### Attributes
- `beta1` (float): The value of `beta1` for moment estimation.
- `beta2` (float): The value of `beta2` for second moment estimation.
- `epsilon` (float): The epsilon value used to avoid division by zero.
- `eta` (float): The learning rate.
- `m` (dict): A dictionary that stores the first moment estimates (mean of gradients) for each parameter.
- `v` (dict): A dictionary that stores the second moment estimates (variance of gradients) for each parameter.

## Methods

### `__init__(self, eta=0.01, beta1=0.9, beta2=0.999, epsilon=1e-8)`
The constructor initializes the Adam optimizer with the given hyperparameters.

### `initialize(self, params)`
This method initializes the first and second moment estimates (`m` and `v`) for each parameter in the provided `params`.

#### Parameters:
- `params` (dict): A dictionary where the keys are parameter names, and the values are the parameters (usually numpy arrays or matrices).

#### Returns:
- None (This modifies the `m` and `v` attributes of the optimizer).

### `update(self, t, params, grads)`
This method updates the model parameters using the Adam optimization rule. It uses the first and second moment estimates to adjust the parameter values.

#### Parameters:
- `t` (int): The current time step or iteration of the optimization.
- `params` (dict): A dictionary of model parameters (keys are parameter names, and values are numpy arrays).
- `grads` (dict): A dictionary of gradients (keys are parameter names, and values are numpy arrays representing the gradients of the respective parameters).

#### Returns:
- `params` (dict): The updated parameters after applying the Adam update.

### Update Rule
The parameters are updated based on the following formulas:
- `m_t = beta1 * m_(t-1) + (1 - beta1) * grad_t`
- `v_t = beta2 * v_(t-1) + (1 - beta2) * (grad_t^2)`
- Correct the bias:
  - `m_hat_t = m_t / (1 - beta1^t)`
  - `v_hat_t = v_t / (1 - beta2^t)`
- Update the parameters:
  - `param_t = param_(t-1) - eta * m_hat_t / (sqrt(v_hat_t) + epsilon)`

### Example Usage

```python
# Initialize the Adam optimizer
adam_optimizer = Adam(eta=0.001)

# Initialize model parameters (example)
params = {
    'w1': np.random.randn(3, 3),
    'b1': np.random.randn(3)
}

# Initialize the optimizer with the model parameters
adam_optimizer.initialize(params)

# Example gradient for each parameter
grads = {
    'w1': np.random.randn(3, 3),
    'b1': np.random.randn(3)
}

# Update parameters at each step
t = 1  # The current time step
params = adam_optimizer.update(t, params, grads)

print(params)
```

### Notes:
- The optimizer must be initialized with the model parameters before using it.
- The `update` method should be called after every iteration (step) in the optimization process.
```

````python
import numpy as np
class Adam():
    def __init__(self, eta=0.01, beta1=0.9, beta2=0.999, epsilon=1e-8):
        
        self.beta1 = beta1
        self.beta2 = beta2
        self.epsilon = epsilon
        self.eta = eta

        self.m = {}
        self.v = {}
    
    def initialize(self, params):

        for key in params:
            self.m[key] = np.zeros_like(params[key])
            self.v[key] = np.zeros_like(params[key])

    def update(self, t, params, grads):
        for key in params:
            self.m[key] = self.beta1 * self.m[key] + (1 - self.beta1) * grads[key]

            self.v[key] = self.beta2 * self.v[key] + (1 - self.beta2) * (grads[key]**2)

            m_corrected = self.m[key] / (1 - self.beta1**t)
            v_corrected = self.v[key] / (1 - self.beta2**t)

            params[key] -= self.eta * m_corrected / (np.sqrt(v_corrected) + self.epsilon)

        return params
````
